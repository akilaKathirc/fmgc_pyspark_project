```python
%sql


create catalog if not exists fmgc;
use catalog fmgc;
```

```python
%sql
create schema if not exists fmgc.bronze;
create schema if not exists fmgc.silver;
create schema if not exists fmgc.gold;
```

```python
external_path = "abfss://fmgc@stgpysparkscenar.dfs.core.windows.net/bronze"
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

schema = StructType([StructField('customer_code', StringType(), True)
                     , StructField('customer', StringType(), True)
                     , StructField('market', StringType(), True)
                     , StructField('platform', StringType(), True)
                     , StructField('channel', StringType(), True)

                     ])


df_customer = spark.read.option("header", True).schema(schema).csv(f"{external_path}/dim_customers.csv")

display(df_customer)
```

```python
%sql
drop table  if exists  fmgc.bronze.dim_customers;
```

```python

df_customer.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable("fmgc.bronze.dim_customers")
```

```python
%sql
drop table  if exists  fmgc.bronze.dim_gross_price;
```

```python


df_gross_price = spark.read.option("header", True).csv(f"{external_path}/dim_gross_price.csv")

display(df_gross_price)


df_customer.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable("fmgc.bronze.dim_gross_price")
```

```python
%sql
drop table   if exists  fmgc.bronze.dim_products;
```

```python


df_products = spark.read.option("header", True).csv(f"{external_path}/dim_products.csv")

display(df_products)


df_customer.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable("fmgc.bronze.dim_products")
```

```python
%sql
drop table if exists fmgc.bronze.fact_orders;
```

```python


df_fact_orders = spark.read.option("header", True).csv(f"{external_path}/fact_orders.csv")

display(df_fact_orders)


df_customer.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable("fmgc.bronze.fact_orders")
```
