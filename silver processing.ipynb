```python
df_customers = spark.table("fmgc.bronze.dim_customers")
display(df_customers)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

```python

df_duplicates = df_customers.withColumn("customer_dupped", rank().over(Window.partitionBy(col("customer")).orderBy("customer")))
display(df_duplicates)
df_count_cust = df_duplicates.filter(col("customer_dupped") > 1)
display(df_count_cust)
```
